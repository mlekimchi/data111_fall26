## 2. Names

An assignment statement like `a = 2` gives a value a **name**, so you can use it later. (slides 8–9.) Watch closely what happens when values change.

````python
a = 2
````

````python
a
````

````python
# The line below causes a name error if b isn't assigned yet.
# b
````

````python
b = 3
a * b
````

`b` exists now, and check: `a` is untouched:

````python
a
````

````python
b
````

````python
total = a + b
total
````

````python
a = 10
total
````

Still 5! A name only updates when *you* re-run its assignment. Verify both, then rebuild `total`:

````python
total
````

````python
a
````

````python
total = a + b
total
````

````python
# The line below causes a syntax error
# a + b = total
````

That one *looks* reasonable but isn't. This next one looks backwards but is everyday Python, it *updates* `a`:

````python
a = a + 1
a
````

> **Assignment is not algebra.** Changing `a` did **not** change `total`, a name only updates when *you* run a new assignment statement. This trips everyone up exactly once. 😬

## 3. Functions and Call Expressions

A **call expression** applies a function to arguments: `round(123.456, 1)`: function name, parentheses, inputs. (Anatomy diagram: KL deck, slides 11–12.)

### 📋 Reference Table

| Code | What it does |
|---|---|
| `abs(-5)` | absolute value |
| `min(14, 15)` | smallest of the arguments |
| `round(123.456, 1)` | round to 1 decimal place |
| `round?` | pull up the documentation |
| `import math` | load a module of extra functions |
| `math.sqrt(2)` | a function that lives in a module |

````python
abs(-5)
````

````python
abs(1 - 3)
````

````python
day_temp = 52
night_temp = 47
abs(night_temp - day_temp)
````

`min` picks the smallest argument; `round` trims decimals, and takes options:

````python
min(14, 15)
````

````python
round(123.456)
````

````python
round(123.456, 1)
````

Arguments can go by *name*, and when you forget the details, `?` pulls up the docs:

````python
round(123.456, ndigits=1)
````

````python
round?
````

````python
round(number=123.456)
````

## 4. Tables 🍦

Enough warm-up! Time to meet the star of the course. A **Table** has labeled columns and one row per record. Our first one is small and delicious.

### 📋 Reference Table

| Code | What it does |
|---|---|
| `Table.read_table('file.csv')` | load a table from a file |
| `t.show(3)` | display the first 3 rows |
| `t.select('A', 'B')` | new table with only those columns |
| `t.drop('A')` | new table without that column |
| `t.where('A', value)` | only rows where column A has that value |
| `t.sort('A')` | sort rows by column A, smallest first |
| `t.sort('A', descending=True)` | sort largest first |

````python
from datascience import *
````

````python
# 🍦

cones = Table.read_table('cones.csv')
cones
````

`show` controls how much prints:

````python
cones.show(3)
````

````python
cones.show()
````

**Choosing columns:** `select` keeps, `drop` removes, and spelling matters:

````python
cones.select('Flavor')
````

````python
cones.select('Flavor', 'Price')
````

````python
# The line below causes an error because Flavor is not a name.
# cones.select(Flavor, 'Price')
````

`drop` is `select`'s mirror image, and notice `cones` itself never changes:

````python
cones.drop('Price')
````

````python
cones
````

````python
cones_without_price = cones.drop('Price')
cones_without_price
````

> **Table methods return a *new* table**: `cones` itself never changed. To keep a result, give it a name: `cones_without_price = cones.drop('Price')`.

````python
cones.where('Flavor', 'chocolate')
````

````python
cones.where('Price',5.25)
````

**Sorting**, three ways:

````python
cones.sort('Price')
````

````python
cones.sort('Price', descending=True)
````

````python
cones.sort('Flavor', descending=True)
````

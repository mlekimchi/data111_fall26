## 2. Defining Functions

You've called `abs`, `round`, `np.average`… now build your own with `def`.

### 📋 Reference Table

| Piece | Example |
|---|---|
| `def` statement | `def triple(x):` |
| body (indented!) | `    return 3 * x` |
| `return` | sends the result back out |
| call it | `triple(4)` → `12` |
| arguments can be expressions | `triple(num * 5)` |

Example: Create a function that takes a numerical input and triples it: $\textsf{triple}(x)=3\,x$

````python
def triple(x):
    return 3 * x
````

````python
triple(3)
````

We can also assign a value to a name, and call the function on the name:

````python
num = 4
````

````python
triple(num)
````

````python
triple(num * 5)
````

## 4. Multiple Arguments

Functions can take as many inputs as you need, just separate them with commas.

Example: Calculate the Hypotenuse Length of a Right Triangle


Pythagoras's Theorem: If $x$ and $y$ denote the lengths of the right-angle sides, then the hypotenuse length $h$ satisfies:

$$ h^2 = x^2 + y^2 \qquad \text{which implies}\qquad \hspace{20 pt} h = \sqrt{ x^2 + y^2 } $$

````python
def hypotenuse(x, y):
    hypot_squared = (x ** 2 + y ** 2)
    hypot = hypot_squared ** 0.5
    return hypot
````

````python
hypotenuse(1, 2)
````

````python
hypotenuse(3, 4)
````

### Example: A function that takes the year of birth of a person and produces their age in years.

````python
def age(year):
    age = 2025 - year
    return age
````

````python
age(1982)
````

Now add some bells and whistles:  Take person's name and year of birth (two arguments). Produce a sentence that states how old they are.

````python
def name_and_age(name, year):
    return name + ' is ' + str(age(year)) + ' years old.'
````

````python
name_and_age('Joe', 1942)
````

## 5. `apply` ✨

Here's the payoff: `t.apply(f, 'col')` calls your function on **every row's value** and hands back an array, no repetition, no manual loops.

### 📋 Reference Table

| Code | What it does |
|---|---|
| `t.apply(f, 'col')` | call `f` on each value in a column |
| `t.apply(f, 'col1', 'col2')` | two-argument functions: one value from each column, row by row |
| result | always an array — attach it with `with_column` |

````python
ages = Table().with_columns(
    'Person', make_array('Jim', 'Pam', 'Michael', 'Creed'),
    'Birth Year', make_array(1985, 1988, 1967, 1904)
)
ages
````

````python
ages.apply(age, 'Birth Year')
````

That one line replaced all of this manual labor:

````python
make_array(age(ages.column('Birth Year').item(0)),
           age(ages.column('Birth Year').item(1)),
           age(ages.column('Birth Year').item(2)),
           age(ages.column('Birth Year').item(3)))
````

````python
age(ages.column('Birth Year'))
````

`age` happens to work on a whole array (it's just arithmetic), but string-building functions don't. That's exactly why `apply` exists:

````python
ages.apply(name_and_age, 'Person', 'Birth Year')
````

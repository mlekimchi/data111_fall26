## 3. Conditional Statements

`if` runs code only when a condition is true; `elif` and `else` handle the other cases. (KL deck, slides 8–9.) Let's build up a dice game one branch at a time: if my roll is bigger you pay me $1, if yours is bigger I pay you $1, ties are a wash.

````python
x = 20
````

````python
if x >= 18:
    print('You can legally vote.')
````

````python
if x >= 21:
    print('You can legally drink.')
````

Let's play a game: we each roll a die. 

If my number is bigger: you pay me a dollar.

If they're the same: we do nothing.

If your number is bigger: I pay you a dollar.

````python
# Work in progress
def one_round(my_roll, your_roll):
    if my_roll > your_roll:
        return 1
````

````python
one_round(4, 3)
````

````python
one_round(2, 6)
````

A losing roll returns… nothing at all. `elif` handles the second case:

````python
# Work in progress
def one_round(my_roll, your_roll):
    if my_roll > your_roll:
        return 1
    elif your_roll > my_roll:
        return -1
````

````python
one_round(2,6)
````

````python
one_round(3,3)
````

## 5. Back to the Dice Game — and Growing Arrays

Instead of making up rolls, let the computer roll. Then, to keep score across many games, we need an array that **grows**: that's `np.append`.

````python
die_faces = np.arange(1, 7)
````

````python
np.random.choice(die_faces)
````

Wrap one full game (random rolls in, score out):

````python
def simulate_one_round():
    my_roll = np.random.choice(die_faces)
    your_roll = np.random.choice(die_faces)
    return one_round(my_roll, your_roll)
````

````python
simulate_one_round()
````

## 6. `for` Statements

Running that cell by hand 10,000 times would take a while. 😅 A `for` loop repeats the body once per element, this is the single most important pattern of the next ten lectures (KL deck, slides 14–16):

```python
for i in np.arange(repetitions):
    ...simulate one outcome, append it...
```

````python
for pet in make_array('cat', 'dog', 'rabbit'):
    print('I love my ' + pet)
````

````python
pet = make_array('cat', 'dog', 'rabbit').item(0)
print('I love my ' + pet)

pet = make_array('cat', 'dog', 'rabbit').item(1)
print('I love my ' + pet)

pet = make_array('cat', 'dog', 'rabbit').item(2)
print('I love my ' + pet)
````

Now the real thing: 5 games, then 10,000:

````python
game_outcomes = make_array()

for i in np.arange(5):
    game_outcomes = np.append(game_outcomes, simulate_one_round())
    
game_outcomes
````

````python
game_outcomes = make_array()

for i in np.arange(10000):
    game_outcomes = np.append(game_outcomes, simulate_one_round())
    
game_outcomes
````

````python
len(game_outcomes)
````

Table it, group it, chart it:

````python
results = Table().with_column('My winnings', game_outcomes)
results
````

````python
results.group('My winnings').barh('My winnings')
````

> **Read that bar chart:** in 10,000 simulated games the three outcomes came out nearly even, exactly what the symmetry of the game predicts. Simulation just *verified* a theory. 🧪

And here's the full simulation recipe: **define one outcome, repeat, collect, visualize.** You'll use this exact skeleton in Lab 5 and Homework 6:

````python
# Simulate one outcome

def num_heads():
    return sum(np.random.choice(coin, 100) == 'heads')
````

````python
# Decide how many times you want to repeat the experiment

repetitions = 10000
````

````python
# Simulate that many outcomes

outcomes = make_array()

for i in np.arange(repetitions):
    outcomes = np.append(outcomes, num_heads())
````

````python
heads = Table().with_column('Heads', outcomes)
heads.hist(bins = np.arange(29.5, 70.6))
````

## 7. Optional: Advanced `where`

`where` also accepts a whole array of bools, or a function, instead of a predicate.

````python
ages = make_array(16, 22, 18, 15, 19, 15, 16, 21)
age = Table().with_column('Age', ages)
````

````python
age
````

````python
age.where('Age', are.above_or_equal_to(18))
````

Now the bool-array way, the comparison itself is the filter:

````python
voter = ages >= 18
````

````python
voter
````

````python
age.where(voter)
````

Or hand `where` a *function* that returns bools:

````python
def my_voter_function(x):
    return x >= 18
````

````python
age.where('Age', my_voter_function)
````

### More Conditional Statements

One more `if` in the wild: classifying every bike trip as round trip or one way, with `apply` + `pivot` to finish.

````python
trip = Table().read_table('trip.csv')
trip.show(3)
````

````python
def trip_kind(start, end):
    if start == end:
        return 'round trip'
    else:
        return 'one way'

kinds = trip.with_column('Trip Kind', trip.apply(trip_kind, 'Start Station', 'End Station'))
kinds.show(3)
````

````python
kinds.where('Duration', are.below(600)).pivot('Trip Kind', 'Start Station')
````
